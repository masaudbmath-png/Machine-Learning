```python
# importing labraries
import pandas as pd
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np
import matplotlib.pyplot as plt
```

```python
file_path='/content/drive/MyDrive/Colab Notebooks/Scikit Learn/Linear Regression/headbrain.csv'
data=pd.read_csv(file_path)
```

```python
data.head()
```

## divid data into feature and target
we will take the data into x and y varaible same like dependent and independent varaibles

```python
x=data['Head Size(cm^3)']
y=data['Brain Weight(grams)']
x.head()
```

```python
y.head()
```

```python
# we must change the dimn of x as
# linearregression need 2 d data
x.shape
```

```python
# this is one dimension data. and linearregression need data in 2 d
x=x.values.reshape(-1,1)
print(x.shape)
print(x.ndim)
```

```python
x_tran, x_test, y_tran, y_test = train_test_split(x,y,test_size=0.10 ,random_state=42)
lr=LinearRegression()
lr.fit(x_tran , y_tran)
```

```python
y_pred=lr.predict(x_test)
```

```python
# to check weather the model is good or not
r2_score(y_test,y_pred)
```

```python
# the r2 score closer to 1 show the model is good in predicting
# lets graph it.
plt.scatter(x_test,y_test, color='blue', label='Real data point')
plt.plot(x_test,y_pred, color='red', label='Predicated values')
plt.xlabel('Head Size')
plt.ylabel('Brain weight')
plt.title('Head Size Vs Brain Weight')
plt.legend()
plt.show
```
