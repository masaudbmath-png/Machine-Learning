# Support Vector Machines (SVM)
Support Vector Machines (SVM) is a `supervised machine learning algorithm` that can be used for both classification and regression tasks. It works by finding the optimal hyperplane that separates data points of different classes in a high-dimensional space. The main goal of SVM is to maximize the margin between the classes, which helps improve the model's generalization ability.

### uses of SVM
1. **Classification**: SVM is widely used for binary and multi-class classification problems. It can effectively handle high-dimensional data and is particularly useful when the number of features exceeds the number of samples.
2. **Regression**: SVM can also be used for regression tasks, known as Support Vector Regression (SVR). It aims to find a function that approximates the relationship between input features and continuous output values.
3. **Outlier Detection**: SVM can be employed for anomaly detection by identifying data points that deviate significantly from the majority of the data. This is useful in applications such as fraud detection and network security.
4. **Text Classification**: SVM is commonly used in natural language processing tasks, such as spam detection, sentiment analysis, and document categorization. It can effectively handle large text datasets and high-dimensional feature spaces.
5. **Image Classification**: SVM can be applied to image recognition tasks, where it can classify images based on their features. It is particularly effective when combined with feature extraction techniques like Histogram of Oriented Gradients (HOG) or Scale-Invariant Feature Transform (SIFT).
6. **Bioinformatics**: SVM is used in bioinformatics for tasks such as protein classification, gene expression analysis, and disease prediction. It can handle complex biological data and identify patterns that may not be apparent through traditional statistical methods.
7. **Face Detection**: SVM can be utilized in computer vision applications for face detection. By training on labeled datasets of facial images, SVM can learn to distinguish between faces and non-faces, making it useful in security systems and social media applications.
8. **Handwriting Recognition**: SVM can be applied to recognize handwritten characters and digits. It is particularly effective in optical character recognition (OCR) systems, where it can classify different handwriting styles and improve accuracy in digitizing handwritten documents.

In this algorithm, we plot each data item as a point in n-dimensional space (where n is the number of features). The value of each feature is then used to determine the position of the point in the space. SVM finds the hyperplane that best separates the classes by maximizing the margin between them. The support vectors are the data points that are closest to the hyperplane and play a crucial role in defining its position.

### Mathematical Formulation of SVM
The mathematical formulation of SVM involves finding the optimal hyperplane that separates the data points of different classes. The hyperplane can be defined by the equation:

$$\mathbf{w} \cdot \mathbf{x} + b = 0$$

where $\mathbf{w}$ is the weight vector, $\mathbf{x}$ is the input feature vector, and $b$ is the bias term.

The goal of SVM is to maximize the margin between the classes, which can be expressed as:
$$\text{Margin} = \frac{2}{||\mathbf{w}||}$$

we need to understand that the margin is defined as the distance between the hyperplane and the closest data points from each class, known as support vectors. The optimization problem can be formulated as:
Minimize: $\frac{1}{2}||\mathbf{w}||^2$
Subject to: $y_i(\mathbf{w} \cdot \mathbf{x}_i + b) \geq 1, \forall i$

### Kernel Trick

The kernel trick is a method used in SVM to handle non-linearly separable data. It allows the algorithm to operate in a higher-dimensional space without explicitly computing the coordinates of the data in that space. Instead, it computes the inner products between the images of pairs of data points in the higher-dimensional space using a kernel function. Common kernel functions include the polynomial kernel and the radial basis function (RBF) kernel.

```python
# importing libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
# importing ml libraries
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from sklearn.preprocessing import StandardScaler,LabelEncoder
```

```python
df=sns.load_dataset('iris')
df.head()
```

```python
# lets make x and y in our data
x=df.drop('species',axis=1)
y=df['species']
```

```python
# train test split the data
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
# call the model
model=SVC()   # we can add perameter inside the brakets like (kernel='rbf)
# train the model
model.fit(x_train,y_train)
#predict the model
pred=model.predict(x_test)
```

```python
# evaluate the model
print(accuracy_score(y_test,pred))
print(confusion_matrix(y_test,pred))
print(classification_report(y_test,pred))
```

```python
# draw the confusion matrxi using heatmap
plt.figure(figsize=(7,4))
sns.heatmap(confusion_matrix(y_test,pred),annot=True)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```
